```python
df.columns = [
    "ID",
    "Name",
    "Quiz",
    "Midterm",
    "Final_Term",
    "Final_Score",
    "Grade"
]

df.head()
```

```python
df.describe()
```

```python
df["Final_Score_Hitung"] = (
    df["Quiz"] * 0.32 +
    df["Midterm"] * 0.33 +
    df["Final_Term"] * 0.35
)

df[["Name", "Quiz", "Midterm", "Final_Term", "Final_Score", "Final_Score_Hitung"]]
```

```python
df["Selisih"] = (
    df["Final_Score"] - df["Final_Score_Hitung"]
).round(2)

df[["Name", "Final_Score", "Final_Score_Hitung", "Selisih"]]
```

```python
mean = df["Final_Score"].mean()

print(f"Mean = {mean:.2f}")
```

```python
median = df["Final_Score"].median()

print(f"Median = {median:.2f}")
```

```python
modus = df["Final_Score"].mode()

print("Modus:")
print(modus)
```

```python
minimum = df["Final_Score"].min()

print(f"Nilai minimum = {minimum:.2f}")
```

```python
maximum = df["Final_Score"].max()

print(f"Nilai maksimum = {maximum:.2f}")
```

```python
range_nilai = maximum - minimum

print(f"Range = {range_nilai:.2f}")
```

```python
variance = df["Final_Score"].var()

print(f"Variance = {variance:.2f}")
```

```python
std_dev = df["Final_Score"].std()

print(f"Standard Deviation = {std_dev:.2f}")
```

```python
q1 = df["Final_Score"].quantile(0.25)

print(f"Q1 = {q1:.2f}")
```

```python
q2 = df["Final_Score"].quantile(0.50)

print(f"Q2 = {q2:.2f}")
```

```python
q3 = df["Final_Score"].quantile(0.75)

print(f"Q3 = {q3:.2f}")
```

```python
iqr = q3 - q1

print(f"IQR = {iqr:.2f}")
```

```python
iqr = q3 - q1

print(f"IQR = {iqr:.2f}")
```

```python
batas_bawah = q1 - (1.5 * iqr)

print(f"Batas bawah = {batas_bawah:.2f}")
```

```python
batas_atas = q3 + (1.5 * iqr)

print(f"Batas atas = {batas_atas:.2f}")
```

```python
outlier = df[
    (df["Final_Score"] < batas_bawah) |
    (df["Final_Score"] > batas_atas)
]

outlier
```

```python
if outlier.empty:
    print("Tidak ditemukan outlier berdasarkan metode 1,5 × IQR.")
else:
    print("Ditemukan outlier:")
    print(outlier[["Name", "Final_Score"]])
```

```python
n = df["Final_Score"].count()

standard_error = std_dev / np.sqrt(n)

print(f"Jumlah data = {n}")
print(f"Standard Error = {standard_error:.2f}")
```

```python
skewness = df["Final_Score"].skew()

print(f"Skewness = {skewness:.3f}")
```

```python
kurtosis = df["Final_Score"].kurt()

print(f"Kurtosis = {kurtosis:.3f}")
```

## Pembahasan

Berdasarkan hasil analisis terhadap 31 data mahasiswa, diperoleh rata-rata nilai akhir sebesar 76,85 dengan median sebesar 79,30. Nilai minimum adalah 46,01, sedangkan nilai maksimum sebesar 95,82. Dengan demikian, range nilai sebesar 49,81 menunjukkan adanya perbedaan capaian antara mahasiswa dengan nilai terendah dan tertinggi.

Nilai Q1 sebesar 68,80 dan Q3 sebesar 87,47 menunjukkan bahwa 50% data tengah berada pada rentang tersebut. Nilai IQR sebesar 18,67. Berdasarkan metode 1,5 × IQR, batas bawah dan batas atas masih mencakup nilai minimum dan maksimum, sehingga tidak ditemukan indikasi outlier.

Standar deviasi sebesar 13,70 menunjukkan adanya variasi nilai antar mahasiswa. Mean yang lebih rendah daripada median juga sejalan dengan nilai skewness sebesar -0,652 yang menunjukkan bahwa distribusi nilai cenderung sedikit menceng ke kiri.

Kurtosis sebesar -0,514 menunjukkan bahwa bentuk distribusi sedikit lebih datar dibandingkan distribusi normal berdasarkan nilai excess kurtosis. Sementara itu, standard error sekitar 2,46 menunjukkan tingkat ketelitian estimasi rata-rata sampel.

Berdasarkan distribusi grade, grade A merupakan grade dengan frekuensi terbesar. Secara keseluruhan, data menunjukkan bahwa capaian mahasiswa berada pada rentang menengah hingga tinggi, meskipun tetap terdapat variasi kemampuan antar mahasiswa.

## Kesimpulan

Berdasarkan analisis statistika deskriptif terhadap 31 mahasiswa, diperoleh mean nilai akhir sebesar 76,85 dan median sebesar 79,30. Nilai akhir memiliki rentang dari 46,01 sampai 95,82 dengan standar deviasi sebesar 13,70.

Dari distribusi grade, grade A memiliki frekuensi paling banyak. Secara umum, hasil analisis menunjukkan adanya variasi nilai antar mahasiswa, tetapi tidak ditemukan pencilan ekstrem berdasarkan metode IQR.
